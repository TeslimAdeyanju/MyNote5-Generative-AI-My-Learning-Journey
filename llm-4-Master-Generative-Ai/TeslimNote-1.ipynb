### Tokenization and Preprocessing

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
# sample text
text1 = """

Think of Machine Learning (ML) as teaching a computer to learn from information. 
It’s like training a baby to recognize a ball by showing it many examples of balls. 
The baby eventually learns to pick the ball from other objects. 
ML is similar, but instead of training a human, we are training a computer, and instead of balls, we give it data to learn from. 
Another important thing is the quality of the data we provide to the ML.

"""
```

```python
print(text1)
print(type(text1))
```

```python
# word tokenization
word_tokenize(text1)
```

```python
text2= """

In this post, we covered:

	1.	Machine Learning teaches computers to learn from data, just like training a baby with examples.
	2.	Features like a customer’s shopping habits help predict their behavior.
	3.	We split data into training, test, and validation sets so the computer learns and is tested correctly.
	4.	A model is like a format that helps the computer make decisions.
	5.	If the model makes errors, we optimize it to improve its predictions.
 
 """
```

```python

print(text2)
print(type(text2))
```

```python
# sentence tokenization
word_tokenize(text2)
```

```python
# sentence tokenization
sent_tokenize(text2)
```

### Stemming and Lemmatization

```python
from nltk.stem import PorterStemmer

# initiate the stemmer
stemmer = PorterStemmer()
```

```python
sample = ['running', 'walking', 'eating', 'worse', 'bigger', 'programming', 'dancing', 'sleeping']
print(sample)
```

Using the code alone before using it as a list function 

```python
stemmer.stem(sample[7])
```

```python
stemmed_words = [stemmer.stem(x) for x in sample]
print(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize(sample[7])
```

```python
lemmatized_words = [lemmatizer.lemmatize(x) for x in sample]
print(lemmatized_words)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv('spam.csv')
```

```python
df
```

```python
df.shape
```

```python
df.value_counts()
```

```python
df['Category'].value_counts()
df['Category'].value_counts(normalize=True)*100
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
    
    
    
df['spam'] = le.fit_transform(df['Category'], )

df.drop(['Category'], axis=1, inplace=True)
```

```python
df
```

### Split the data 

```python
X_train, X_test, y_train, y_test = train_test_split(df['Message'], df['spam'], test_size=0.2, random_state=42)
```

```python
X_train
```

```python

from sklearn.feature_extraction.text import CountVectorizer
# Create an instance of CountVectorizer
v = CountVectorizer()
```

```python
X_train_cv = v.fit_transform(X_train.values)
```

```python
X_test_cv = v.transform(X_test)
```

```python
X_train_cv
```

```python
# Get feature names (tokens)
print(v.get_feature_names_out())
```

```python
X_train_cv.shape
```

```python
v.vocabulary_
```

### Train the model 

```python
model = MultinomialNB()
model.fit(X_train_cv, y_train)
```

```python
y_pred = model.predict(X_test_cv)
```

```python
print(classification_report(y_test, y_pred))
```

```python
message = {"Upto 100% off on parking, exclusing offer just for you"}
```

```python
message_cnt = v.transform(message)

model.predict(message_cnt)
```

```python
# Find words most similar to the combination of these words
a = model.most_similar(positive=['Nigeria', 'corruption', 'government'], topn=10)
```

```python
# Extract similarity scores from the list of tuples and find the maximum
max_similarity = max(score for word, score in a)
print(f"Maximum similarity score: {max_similarity}")

# If you also want to know which word has the maximum score
max_word, max_score = max(a, key=lambda item: item[1])
print(f"Word with maximum similarity: {max_word} ({max_score})")
```
